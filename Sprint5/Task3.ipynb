Создать базовый класс Product в C#, который будет представлять информацию о
продуктах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.


<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

public delegate void ProductEventHandler(string message);

public class Product
{
    public string Name { get; set; }
    public double Price { get; set; }
    public string Manufacturer { get; set; }
    public int StockQuantity { get; set; }
    
    // Новые атрибуты
    public string Country { get; set; }
    public bool IsAvailable { get; set; }
    public DateTime CreatedDate { get; set; }
    
    // Событие
    public event ProductEventHandler ProductUpdated;

    public Product(string name, double price, string manufacturer, int stockQuantity)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        StockQuantity = stockQuantity;
        CreatedDate = DateTime.Now;
        UpdateAvailability();
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}, В наличии: {StockQuantity} шт.");
    }

    public virtual void Discount(double percent)
    {
        if (percent > 0 && percent <= 100)
        {
            Price -= Price * (percent / 100);
            ProductUpdated?.Invoke($"Применена скидка {percent}% на {Name}");
        }
    }

    // Новые методы
    public void SetCountry(string country)
    {
        Country = country;
        ProductUpdated?.Invoke($"Страна производства {Name} установлена: {country}");
    }
    
    public void UpdateAvailability()
    {
        IsAvailable = StockQuantity > 0;
    }
    
    public void Sell(int quantity)
    {
        if (StockQuantity >= quantity)
        {
            StockQuantity -= quantity;
            UpdateAvailability();
            ProductUpdated?.Invoke($"Продано {quantity} шт. товара {Name}");
        }
    }
}

public class Electronics : Product
{
    // Новые атрибуты
    public int WarrantyPeriod { get; set; }
    public string Model { get; set; }
    public bool HasBattery { get; set; }
    public string Color { get; set; }

    public Electronics(string name, double price, string manufacturer, int stockQuantity, 
                      int warrantyPeriod, string model, bool hasBattery, string color) 
        : base(name, price, manufacturer, stockQuantity)
    {
        WarrantyPeriod = warrantyPeriod;
        Model = model;
        HasBattery = hasBattery;
        Color = color;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Гарантия: {WarrantyPeriod} мес., Модель: {Model}, Цвет: {Color}");
    }

    public override void Discount(double percent)
    {
        double maxDiscount = Math.Min(percent, 25);
        base.Discount(maxDiscount);
    }

    // Новые методы
    public void TestDevice()
    {
        Console.WriteLine($"Тестирование устройства {Name}...");
    }
    
    public void CheckBattery()
    {
        Console.WriteLine($"Батарея: {HasBattery}");
    }
}

public class Clothing : Product
{
    // Новые атрибуты
    public string Size { get; set; }
    public string Material { get; set; }
    public string Season { get; set; }
    public string Gender { get; set; }

    public Clothing(string name, double price, string manufacturer, int stockQuantity, 
                   string size, string material, string season, string gender) 
        : base(name, price, manufacturer, stockQuantity)
    {
        Size = size;
        Material = material;
        Season = season;
        Gender = gender;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размер: {Size}, Материал: {Material}, Сезон: {Season}");
    }

    // Новые методы
    public void TryOn()
    {
        Console.WriteLine($"Примерка: {Name}, размер {Size}");
    }
    
    public void Wash()
    {
        Console.WriteLine($"Стирка одежды {Name}");
    }
}

public class Books : Product
{
    // Новые атрибуты
    public string Author { get; set; }
    public string Genre { get; set; }
    public int PageCount { get; set; }
    public string Language { get; set; }

    public Books(string name, double price, string manufacturer, int stockQuantity, 
                string author, string genre, int pageCount, string language) 
        : base(name, price, manufacturer, stockQuantity)
    {
        Author = author;
        Genre = genre;
        PageCount = pageCount;
        Language = language;
    }

    public override void DisplayInfo()
    {
        Console.WriteLine($"Книга: {Name}, Автор: {Author}, Жанр: {Genre}, Страниц: {PageCount}, Цена: {Price} руб.");
    }

    // Новые методы
    public void ReviewBook(int rating)
    {
        Console.WriteLine($"Оценка книги '{Name}': {rating}/5 звезд");
    }
    
    public void ReadSample()
    {
        Console.WriteLine($"Чтение отрывка из книги '{Name}'...");
    }
}


static void OnProductUpdated(string message)
{
Console.WriteLine($"[СОБЫТИЕ] {message}");
}

{
// Создание продуктов
Product laptop = new Electronics("Ноутбук", 50000, "Lenovo", 5, 24, "ThinkPad", true, "Черный");
Product tshirt = new Clothing("Футболка", 1500, "Nike", 10, "L", "Хлопок", "Лето", "Унисекс");
Product book = new Books("Война и мир", 800, "Эксмо", 3, "Лев Толстой", "Роман", 1225, "Русский");

// Подписка на события
laptop.ProductUpdated += OnProductUpdated;
tshirt.ProductUpdated += OnProductUpdated;
book.ProductUpdated += OnProductUpdated;

// Использование List<T>
List<Product> products = new List<Product> { laptop, tshirt, book };

Console.WriteLine("=== ВСЕ ПРОДУКТЫ ===");
products.ForEach(p => p.DisplayInfo());

Console.WriteLine("\n=== ДЕЛЕГАТЫ И ЛЯМБДА-ВЫРАЖЕНИЯ ===");

// Делегаты
Action<Product> displayAction = p => p.DisplayInfo();
Console.WriteLine("Делегат Display:");
displayAction(laptop);

// Лямбда-выражения с коллекциями
var expensiveProducts = products.FindAll(p => p.Price > 1000);
Console.WriteLine("\nДорогие товары (>1000 руб):");
expensiveProducts.ForEach(p => p.DisplayInfo());

var availableProducts = products.FindAll(p => p.IsAvailable);
Console.WriteLine("\nДоступные товары:");
availableProducts.ForEach(p => p.DisplayInfo());

Console.WriteLine("\n=== СОБЫТИЯ И МЕТОДЫ ===");

// Вызов методов, которые вызывают события
laptop.Discount(10);
laptop.SetCountry("Китай");
laptop.Sell(2);

tshirt.Discount(15);
tshirt.Sell(1);

book.Discount(5);
book.Sell(1);


Console.WriteLine("\n=== МЕТОДЫ ПРОИЗВОДНЫХ КЛАССОВ ===");

((Electronics)laptop).TestDevice();
((Electronics)laptop).CheckBattery();

((Clothing)tshirt).TryOn();
((Clothing)tshirt).Wash();

((Books)book).ReviewBook(5);
((Books)book).ReadSample();
}
```
