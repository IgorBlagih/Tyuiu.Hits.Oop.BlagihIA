```
using System;
using System.Collections.Generic;

// Базовый класс Product
public class Product
{
    // Атрибуты
    public string Name { get; set; }
    public double Price { get; set; }
    public string Manufacturer { get; set; }
    public int StockQuantity { get; set; } 
    public string Category { get; set; } 
    public string Barcode { get; set; }
    
    // Новые атрибуты
    public string Country { get; set; }
    public bool IsAvailable { get; set; }
    public DateTime CreatedDate { get; set; }
    
    // Дополнительные атрибуты
    public double Rating { get; set; }
    public int SalesCount { get; set; }
    public bool IsFeatured { get; set; }

    // Конструктор
    public Product(string name, double price, string manufacturer, int stockQuantity, string category, string barcode)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        StockQuantity = stockQuantity;
        Category = category;
        Barcode = barcode;
        CreatedDate = DateTime.Now;
        Rating = 0;
        SalesCount = 0;
        IsFeatured = false;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}, В наличии: {StockQuantity} шт.";
    }
    
    // Перегрузка метода GetInfo
    public string GetInfo(bool detailed)
    {
        if (detailed)
            return $"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}, В наличии: {StockQuantity} шт., Страна: {Country}, Доступен: {IsAvailable}";
        else
            return GetInfo();
    }

    public virtual void Discount(double disPer)
    {
        if (disPer > 0 && disPer <= 100)
        {
            Price -= Price * (disPer / 100);
            Console.WriteLine($"Применена скидка {disPer}%. Новая цена: {Price} руб.");
        }
    }
    
    // Перегрузка метода Discount
    public void Discount(double disPer, string reason)
    {
        Discount(disPer);
        Console.WriteLine($"Причина скидки: {reason}");
    }

    public void Restock(int quantity)
    {
        StockQuantity += quantity;
        Console.WriteLine($"Товар {Name} пополнен на {quantity} шт.");
    }
    
    // Новые методы
    public void SetCountry(string country)
    {
        Country = country;
        Console.WriteLine($"Страна производства установлена: {country}");
    }
    
    public void UpdateAvailability()
    {
        IsAvailable = StockQuantity > 0;
        Console.WriteLine($"Доступность {Name}: {IsAvailable}");
    }
    
    public virtual void DisplayCategory()
    {
        Console.WriteLine($"Категория товара: {Category}");
    }

    public void Sell(int quantity)
    {
        if (StockQuantity >= quantity)
        {
            StockQuantity -= quantity;
            SalesCount += quantity;
            Console.WriteLine($"Продано {quantity} шт. товара {Name}");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара {Name} на складе");
        }
    }

    public void ChangePrice(double newPrice)
    {
        Price = newPrice;
        Console.WriteLine($"Цена товара {Name} изменена на {newPrice} руб.");
    }

    // Метод взаимодействия с другим продуктом
    public void ComparePrice(Product otherProduct)
    {
        if (Price > otherProduct.Price)
        {
            Console.WriteLine($"{Name} дороже чем {otherProduct.Name} на {Price - otherProduct.Price} руб.");
        }
        else if (Price < otherProduct.Price)
        {
            Console.WriteLine($"{Name} дешевле чем {otherProduct.Name} на {otherProduct.Price - Price} руб.");
        }
        else
        {
            Console.WriteLine($"{Name} и {otherProduct.Name} стоят одинаково");
        }
    }
    
    // Дополнительные методы
    public void UpdateRating(double newRating)
    {
        Rating = newRating;
        Console.WriteLine($"Рейтинг товара {Name} обновлен: {newRating}/5");
    }
    
    public bool IsPopular()
    {
        return SalesCount > 100 || Rating >= 4.5;
    }
    
    public void MarkAsFeatured()
    {
        IsFeatured = true;
        Console.WriteLine($"Товар {Name} отмечен как рекомендуемый");
    }
    
    public virtual double CalculateProfit()
    {
        return Price * SalesCount;
    }
}

// Generic класс
public class ProductCollection<T> where T : Product
{
    private List<T> _products = new List<T>();

    public void Add(T product)
    {
        _products.Add(product);
    }

    public void Remove(T product)
    {
        _products.Remove(product);
    }

    public void DisplayAll()
    {
        foreach (var product in _products)
        {
            Console.WriteLine(product.GetInfo());
        }
    }
    
    // Перегрузка
    public void DisplayAll(bool detailed)
    {
        foreach (var product in _products)
        {
            if (detailed)
                Console.WriteLine(product.GetInfo(true));
            else
                Console.WriteLine(product.GetInfo());
        }
    }
    
    public T FindByName(string name)
    {
        return _products.Find(p => p.Name == name);
    }
    
    public List<T> GetFeaturedProducts()
    {
        return _products.FindAll(p => p.IsFeatured);
    }
}

// Интерфейс для товаров с гарантией
public interface IWarranty
{
    int WarrantyPeriod { get; set; }
    void ExtendWarranty(int months);
    bool IsWarrantyValid();
    string GetWarrantyInfo(); // Новый метод
}

// Интерфейс для товаров с доставкой
public interface IDelivery
{
    double DeliveryCost { get; set; }
    void CalculateDelivery();
    string GetDeliveryTime();
    bool CanExpressDelivery(); // Новый метод
}

// Новый интерфейс для скидок
public interface IDiscountable
{
    void ApplySeasonalDiscount();
    bool IsDiscountEligible();
    double GetMaxDiscount();
}

// Производный класс Electronics (множественное наследование)
public class Electronics : Product, IWarranty, IDiscountable
{
    // Атрибуты
    public int WarrantyPeriod { get; set; }
    public string PowerSupply { get; set; } 
    public double Weight { get; set; } 
    public string EnergyClass { get; set; }
    
    // Новые атрибуты
    public string Model { get; set; }
    public bool HasBattery { get; set; }
    public string Color { get; set; }
    
    // Дополнительные атрибуты
    public int YearOfManufacture { get; set; }

    // Конструктор
    public Electronics(string name, double price, string manufacturer, int stockQuantity, 
                      int warrantyPeriod, string powerSupply, double weight, string energyClass) 
        : base(name, price, manufacturer, stockQuantity, "Электроника", Guid.NewGuid().ToString())
    {
        WarrantyPeriod = warrantyPeriod;
        PowerSupply = powerSupply;
        Weight = weight;
        EnergyClass = energyClass;
        YearOfManufacture = DateTime.Now.Year;
    }

    public override void Discount(double disPer)
    {
        double maxDis = 30 - (WarrantyPeriod / 2);
        if (maxDis < 5) maxDis = 5;
        
        if (disPer > maxDis)
        {
            Console.WriteLine($"Максимальная скидка {maxDis}%");
            disPer = maxDis;
        }
        
        base.Discount(disPer);
    }
    
    // Перегрузка
    public void Discount(double disPer, bool isSeasonal)
    {
        if (isSeasonal)
            disPer += 5;
        
        Discount(disPer);
    }

    // Явная реализация интерфейса IWarranty
    void IWarranty.ExtendWarranty(int months)
    {
        WarrantyPeriod += months;
        Console.WriteLine($"Гарантия продлена на {months} мес. (явная реализация)");
    }

    bool IWarranty.IsWarrantyValid()
    {
        return WarrantyPeriod > 0;
    }

    string IWarranty.GetWarrantyInfo()
    {
        return $"Гарантия: {WarrantyPeriod} месяцев, производитель: {Manufacturer}";
    }

    // Явная реализация интерфейса IDiscountable
    void IDiscountable.ApplySeasonalDiscount()
    {
        Discount(15, "сезонная скидка на электронику");
    }

    bool IDiscountable.IsDiscountEligible()
    {
        return YearOfManufacture >= DateTime.Now.Year - 2;
    }

    double IDiscountable.GetMaxDiscount()
    {
        return 30.0;
    }

    public void TestDevice()
    {
        Console.WriteLine($"Тестирование устройства {Name}...");
    }
    
    // Новые методы
    public void SetModel(string model)
    {
        Model = model;
        Console.WriteLine($"Модель установлена: {model}");
    }
    
    public void CheckBattery()
    {
        Console.WriteLine($"Проверка батареи устройства {Name}: {HasBattery}");
    }
    
    public override void DisplayCategory()
    {
        Console.WriteLine($"Электроника: {Name}, Класс энергопотребления: {EnergyClass}");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес., Вес: {Weight}кг, Класс энергопотребления: {EnergyClass}";
    }
    
   
}

//Подкласс TV класса Electronics(сложное наследование)
public class TV : Electronics
{
    // Новые атрибуты
    public int Diagonal { get; set; }
    public string Resolution { get; set; }
    public bool IsSmartTV { get; set; }
    
    // Новые атрибуты
    public bool HasHDR { get; set; }
    public int HDMIports { get; set; }
    public string ScreenType { get; set; }
    
    // Дополнительные атрибуты
    public string OperatingSystem { get; set; }
    public bool HasVoiceControl { get; set; }
    public int RefreshRate { get; set; }
    public string PanelType { get; set; }

    // Конструктор
    public TV(string name, double price, string manufacturer, int stockQuantity, 
              int warrantyPeriod, string powerSupply, double weight, string energyClass,
              int diagonal, string resolution, bool isSmartTV) 
        : base(name, price, manufacturer, stockQuantity, warrantyPeriod, powerSupply, weight, energyClass)
    {
        Diagonal = diagonal;
        Resolution = resolution;
        IsSmartTV = isSmartTV;
        RefreshRate = 60;
        PanelType = "LED";
    }

    // Новые методы
    public void MovieOn()
    {
        Console.WriteLine($"На телевизоре {Name} включена витринная видеозаставка");
    }
    
    // Перегрузка
    public void MovieOn(string movieName)
    {
        Console.WriteLine($"На телевизоре {Name} включен фильм: {movieName}");
    }

    public void ChangeChannel(int channel)
    {
        Console.WriteLine($"Телевизор {Name} переключен на канал {channel}");
    }

    public void AdjustVolume(int level)
    {
        Console.WriteLine($"Громкость телевизора {Name} установлена на уровень {level}");
    }
    

}

// Класс Clothing(простое наследование)
public class Clothing : Product, IDiscountable
{
    public string Size { get; set; }
    public string Color { get; set; } 
    public string Material { get; set; } 
    public string Season { get; set; }
    
    // Новые атрибуты
    public string Gender { get; set; }
    public string CareInstructions { get; set; }
    public bool IsEcoFriendly { get; set; }
    
    // Дополнительные атрибуты
    public string FitType { get; set; }
    public string Collection { get; set; }
    public bool IsSustainable { get; set; }
    public string SizeChart { get; set; }

    // Конструктор
    public Clothing(string name, double price, string manufacturer, int stockQuantity, 
                   string size, string color, string material, string season) 
        : base(name, price, manufacturer, stockQuantity, "Одежда", Guid.NewGuid().ToString())
    {
        Size = size;
        Color = color;
        Material = material;
        Season = season;
        FitType = "Regular";
    }

    public void TryOn()
    {
        Console.WriteLine($"Примерка: {Name}, размер {Size}");
    }
    
    // Перегрузка
    public void TryOn(string location)
    {
        Console.WriteLine($"Примерка: {Name} в {location}");
    }

    public void Wash()
    {
        Console.WriteLine($"Стирка одежды {Name}");
    }

    public void ChangeColor(string newColor)
    {
        Color = newColor;
        Console.WriteLine($"Цвет одежды {Name} изменен на {newColor}");
    }
    
    // Новые методы
    public void SetGender(string gender)
    {
        Gender = gender;
        Console.WriteLine($"Пол установлен: {gender}");
    }
    
    public void ShowCareInstructions()
    {
        Console.WriteLine($"Инструкции по уходу для {Name}: {CareInstructions}");
    }
    
    public override void DisplayCategory()
    {
        Console.WriteLine($"Одежда: {Name}, Размер: {Size}, Материал: {Material}");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Размер: {Size}, Цвет: {Color}, Материал: {Material}, Сезон: {Season}";
    }
    
    // Явная реализация интерфейса IDiscountable
    void IDiscountable.ApplySeasonalDiscount()
    {
        if (Season == "Зима" && DateTime.Now.Month >= 11 || DateTime.Now.Month <= 2)
        {
            Discount(20, "зимняя распродажа");
        }
        else if (Season == "Лето" && DateTime.Now.Month >= 6 && DateTime.Now.Month <= 8)
        {
            Discount(15, "летняя распродажа");
        }
    }

    bool IDiscountable.IsDiscountEligible()
    {
        return StockQuantity > 50;
    }

    double IDiscountable.GetMaxDiscount()
    {
        return 40.0;
    }
    

}

// Производный класс Books (множественное наследование через интерфейсы)
public class Books : Product, IDelivery, IDiscountable
{
    public string Author { get; set; }
    public int PageCount { get; set; } 
    public string Genre { get; set; } 
    public string CoverType { get; set; } 
    public double DeliveryCost { get; set; }
    
    // Новые атрибуты
    public string Language { get; set; }
    public int Edition { get; set; }
    public bool IsBestseller { get; set; }
    
    // Дополнительные атрибуты
    public string Publisher { get; set; }
    public int PublicationYear { get; set; }
    public bool IsIllustrated { get; set; }
    public string ISBN { get; set; }

    // Конструктор
    public Books(string name, double price, string manufacturer, int stockQuantity, 
                string author, int pageCount, string genre, string coverType) 
        : base(name, price, manufacturer, stockQuantity, "Книги", Guid.NewGuid().ToString())
    {
        Author = author;
        PageCount = pageCount;
        Genre = genre;
        CoverType = coverType;
        DeliveryCost = 0;
        PublicationYear = DateTime.Now.Year;
    }
   
    // Явная реализация интерфейса IDelivery
    void IDelivery.CalculateDelivery()
    {
        DeliveryCost = PageCount > 500 ? 300 : 200;
        Console.WriteLine($"Стоимость доставки книги: {DeliveryCost} руб. (явная реализация)");
    }
    
    // Перегрузка
    public void CalculateDelivery(bool express)
    {
        if (express)
            DeliveryCost = 500;
        else
            DeliveryCost = 200;
        Console.WriteLine($"Доставка книги: {DeliveryCost} руб.");
    }

    string IDelivery.GetDeliveryTime()
    {
        return PageCount > 1000 ? "5-7 рабочих дней" : "3-5 рабочих дней";
    }

    bool IDelivery.CanExpressDelivery()
    {
        return StockQuantity > 0 && Price > 1000;
    }

    // Явная реализация интерфейса IDiscountable
    void IDiscountable.ApplySeasonalDiscount()
    {
        if (IsBestseller)
        {
            Discount(10, "скидка на бестселлер");
        }
        else if (PublicationYear < DateTime.Now.Year - 5)
        {
            Discount(25, "скидка на старые издания");
        }
    }

    bool IDiscountable.IsDiscountEligible()
    {
        return !IsBestseller || StockQuantity > 20;
    }

    double IDiscountable.GetMaxDiscount()
    {
        return 50.0;
    }

    public void ReviewBook(int rating)
    {
        Console.WriteLine($"Оценка книги '{Name}': {rating}/5 звезд");
    }
    
    // Новые методы
    public void SetLanguage(string language)
    {
        Language = language;
        Console.WriteLine($"Язык книги установлен: {language}");
    }
    
    public void CheckEdition()
    {
        Console.WriteLine($"Книга '{Name}' - {Edition} издание");
    }
    
    public override void DisplayCategory()
    {
        Console.WriteLine($"Книга: {Name}, Автор: {Author}, Жанр: {Genre}");
    }

    public override string GetInfo()
    {
        return $"Книга: {Name}, Автор: {Author}, Жанр: {Genre}, Страниц: {PageCount}, Обложка: {CoverType}, Цена: {Price} руб.";
    }
    

}

// Сервис для управления продуктами (управление зависимостями)
public class ProductService
{
    private readonly ProductCollection<Product> _productCollection;

    public ProductService(ProductCollection<Product> productCollection)
    {
        _productCollection = productCollection;
    }

    public void ProcessFeaturedProducts()
    {
        var featuredProducts = _productCollection.GetFeaturedProducts();
        Console.WriteLine("\n=== Рекомендуемые товары ===");
        foreach (var product in featuredProducts)
        {
            Console.WriteLine($"★ {product.Name} - {product.Price} руб.");
        }
    }

    public void ApplySeasonalDiscounts()
    {
        Console.WriteLine("\n=== Сезонные скидки ===");
        foreach (var product in _productCollection.GetFeaturedProducts())
        {
            if (product is IDiscountable discountable)
            {
                if (discountable.IsDiscountEligible())
                {
                    discountable.ApplySeasonalDiscount();
                }
            }
        }
    }

    public void CheckDeliveryOptions()
    {
        Console.WriteLine("\n=== Опции доставки ===");
        foreach (var product in _productCollection.GetFeaturedProducts())
        {
            if (product is IDelivery deliverable)
            {
                deliverable.CalculateDelivery();
                Console.WriteLine($"Время доставки: {deliverable.GetDeliveryTime()}");
            }
        }
    }

    public void DisplayProfitReport()
    {
        Console.WriteLine("\n=== Отчет о прибыли ===");
        double totalProfit = 0;
        foreach (var product in _productCollection.GetFeaturedProducts())
        {
            double profit = product.CalculateProfit();
            totalProfit += profit;
            Console.WriteLine($"{product.Name}: {profit} руб.");
        }
        Console.WriteLine($"Общая прибыль: {totalProfit} руб.");
    }
}


// Создание объектов
Product myProduct1 = new Electronics("Ноутбук", 50000, "Lenovo", 5, 24, "220V", 2.1, "A+");
Product myProduct2 = new Clothing("Футболка", 1500, "Nike", 10, "L", "Синий", "Хлопок", "Лето");
Product myProduct3 = new Books("Война и мир", 800, "Эксмо", 3, "Лев Толстой", 1225, "Роман", "Твердая");
Product myTV = new TV("QLED", 75000, "Samsung", 3, 5, "220V", 8, "A++", 45, "4K UHD", true);

// Использование Generic класса
ProductCollection<Product> productCollection = new ProductCollection<Product>();
productCollection.Add(myProduct1);
productCollection.Add(myProduct2);
productCollection.Add(myProduct3);
productCollection.Add(myTV);

// Отмечаем товары как рекомендуемые
myProduct1.MarkAsFeatured();
myProduct2.MarkAsFeatured();
myTV.MarkAsFeatured();

Console.WriteLine("=== Generic класс ===");
productCollection.DisplayAll();

Console.WriteLine("\n=== Управление зависимостями через ProductService ===");
ProductService productService = new ProductService(productCollection);
productService.ProcessFeaturedProducts();
productService.ApplySeasonalDiscounts();
productService.CheckDeliveryOptions();
productService.DisplayProfitReport();

Console.WriteLine("\n=== Явная реализация интерфейсов ===");
// Явная реализация IWarranty
IWarranty warrantyElectronics = (IWarranty)myProduct1;
Console.WriteLine(warrantyElectronics.GetWarrantyInfo());
warrantyElectronics.ExtendWarranty(6);

// Явная реализация IDelivery
IDelivery deliveryBook = (IDelivery)myProduct3;
deliveryBook.CalculateDelivery();
Console.WriteLine($"Время доставки: {deliveryBook.GetDeliveryTime()}");

// Явная реализация IDiscountable
IDiscountable discountableClothing = (IDiscountable)myProduct2;
if (discountableClothing.IsDiscountEligible())
{
    discountableClothing.ApplySeasonalDiscount();
}

Console.WriteLine("\n=== Дополнительные методы ===");
myProduct1.UpdateRating(4.8);


```
