Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

// Базовый класс Product
public class Product
{
    public string Name { get; set; }
    public double Price { get; set; }
    public string Manufacturer { get; set; }

    // Конструктор
    public Product(string name, double price, string manufacturer)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}";
    }

    public virtual void Discount(double disPer)
    {
        if (disPer > 0 && disPer <= 100)
        {
            Price -= Price * (disPer / 100);
            Console.WriteLine($"Применена скидка {disPer}%. Новая цена: {Price} руб.");
        }
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    // Метод взаимодействия с другим продуктом
    public void ComparePrice(Product otherProduct)
    {
        
        if (Price > otherProduct.Price)
        {
            Console.WriteLine($"{Name} дороже чем {otherProduct.Name} на {Price - otherProduct.Price} руб.");
        }
        else if (Price < otherProduct.Price)
        {
            Console.WriteLine($"{Name} дешевле чем {otherProduct.Name} на {otherProduct.Price - Price} руб.");
        }
        else
        {
            Console.WriteLine($"{Name} и {otherProduct.Name} стоят одинаково");
        }
    }
}

// Производный класс Electronics
public class Electronics : Product
{
    public int WarrantyPeriod { get; set; }

    // Конструктор
    public Electronics(string name, double price, string manufacturer, int warrantyPeriod) 
        : base(name, price, manufacturer)
    {
        WarrantyPeriod = warrantyPeriod;
    }

    public override void Discount(double disPer)
    {
        double maxDis = 30 - (WarrantyPeriod / 2);
        if (maxDis < 5) maxDis = 5;
        
        if (disPer > maxDis)
        {
            Console.WriteLine($"Для товара с гарантией {WarrantyPeriod} мес. максимальная скидка {maxDis}%");
            disPer = maxDis;
        }
        
        base.Discount(disPer);
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес.";
    }
}

// Производный класс Clothing
public class Clothing : Product
{
    public string Size { get; set; }

    // Конструктор
    public Clothing(string name, double price, string manufacturer, string size) 
        : base(name, price, manufacturer)
    {
        Size = size;
    }

    public override void Display()
    {
        Console.WriteLine($"Одежда: {Name}, Размер: {Size}, Цена: {Price} руб., Производитель: {Manufacturer}");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Размер: {Size}";
    }
}

// Производный класс Books
public class Books : Product
{
    public string Author { get; set; }

    // Конструктор
    public Books(string name, double price, string manufacturer, string author) 
        : base(name, price, manufacturer)
    {
        Author = author;
    }

    public override string GetInfo()
    {
        return $"Книга: {Name}, Автор: {Author}, Цена: {Price} руб., Издательство: {Manufacturer}";
    }
}

Product myProduct1 = new Electronics("Ноутбук", 50000, "Lenovo", 24);
Product myProduct2 = new Clothing("Футболка", 1500, "Nike", "L");
Product myProduct3 = new Books("Война и мир", 800, "Эксмо", "Лев Толстой");

// Взаимодействие объектов между собой
myProduct1.ComparePrice(myProduct2);
myProduct2.ComparePrice(myProduct3);
myProduct3.ComparePrice(myProduct1);

Console.WriteLine("\nDisplay");
myProduct1.Display();
myProduct2.Display();
myProduct3.Display();

Console.WriteLine("\nGetInfo");
Console.WriteLine(myProduct1.GetInfo());
Console.WriteLine(myProduct2.GetInfo());
Console.WriteLine(myProduct3.GetInfo());

Console.WriteLine("\nDiscount");
myProduct1.Discount(25);
myProduct2.Discount(15);
myProduct3.Discount(10);

Console.WriteLine("\nDisplay после скидок");
myProduct1.Display();
myProduct2.Display();
myProduct3.Display();

Console.WriteLine("\nСравнение цен после скидок:");
myProduct2.ComparePrice(myProduct3);
```
