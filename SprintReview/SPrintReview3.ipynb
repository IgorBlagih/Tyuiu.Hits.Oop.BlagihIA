```
using System;
using System.Collections.Generic;

// Базовый абстрактный класс Component
public abstract class Component
{
    public string Name { get; set; }
    public double Price { get; set; }
    public string Manufacturer { get; set; }

    public Component(string name, double price, string manufacturer)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
    }

    // виртуальный метод GetInfo
    public virtual string GetInfo()
    {
        return $"{Name} - {Manufacturer} - {Price} руб.";
    }

    // перегрузка ToString
    public override string ToString()
    {
        return GetInfo();
    }
}

// интерфейс IInstallable
public interface IInstallable
{
    void Install();
}

// класс CPU
public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double Frequency { get; set; }

    public CPU(string name, double price, string manufacturer, int cores, double frequency) 
        : base(name, price, manufacturer)
    {
        Cores = cores;
        Frequency = frequency;
    }

    // перекрытие метода
    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Ядер: {Cores}, Частота: {Frequency} GHz";
    }

    // реализация интерфейса
    public void Install()
    {
        Console.WriteLine($"- Установка процессора {Name} в сокет");
    }
}

// класс RAM
public class RAM : Component, IInstallable
{
    public int Capacity { get; set; }
    public string Type { get; set; }

    public RAM(string name, double price, string manufacturer, int capacity, string type) 
        : base(name, price, manufacturer)
    {
        Capacity = capacity;
        Type = type;
    }

    // Перекрытие метода
    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Память: {Capacity} GB, Тип: {Type}";
    }

    // Реализация интерфейса
    public void Install()
    {
        Console.WriteLine($"- Установка памяти {Name} в слот DIMM");
    }
}

// Класс GPU
public class GPU : Component, IInstallable
{
    public int Memory { get; set; }
    public string Interface { get; set; }

    public GPU(string name, double price, string manufacturer, int memory, string interfaceType) 
        : base(name, price, manufacturer)
    {
        Memory = memory;
        Interface = interfaceType;
    }

    // перекрытие метода
    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Видеопамять: {Memory} GB, Интерфейс: {Interface}";
    }

    // реализация интерфейса
    public void Install()
    {
        Console.WriteLine($"- Установка видеокарты {Name} в PCI-E слот");
    }
}

// класс HDD
public class HDD : Component, IInstallable
{
    public int Capacity { get; set; }
    public string FormFactor { get; set; }

    public HDD(string name, double price, string manufacturer, int capacity, string formFactor) 
        : base(name, price, manufacturer)
    {
        Capacity = capacity;
        FormFactor = formFactor;
    }

    // перекрытие метода
    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Объем: {Capacity} GB, Форм-фактор: {FormFactor}";
    }

    // реализаци интерфейса
    public void Install()
    {
        Console.WriteLine($"- Установка диска {Name} в отсек");
    }
}

// класс Computer с Generic методами
public class Computer
{
    private List<Component> components = new List<Component>();

    // Generic метод для добавления компонента
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();
        Console.WriteLine($"- Компонент {component.Name} добавлен в компьютер");
    }

    // Generic метод для удаления компонента
    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
        {
            Console.WriteLine($"- Компонент {component.Name} удален из компьютера");
        }
    }

    // метод для получения общей стоимости
    public double GetTotalPrice()
    {
        double total = 0;
        foreach (var component in components)
        {
            total += component.Price;
        }
        return total;
    }

    // метод для отображения всех компонентов
    public void ShowComponents()
    {
        Console.WriteLine("\nКомпоненты компьютера:");
        foreach (var component in components)
        {
            Console.WriteLine($" - {component.GetInfo()}");
        }
        Console.WriteLine($"\nОбщая стоимость: {GetTotalPrice()} руб.");
    }
}

// статический класс фабрики
public static class ComputerFactory
{
    public static Computer PC()
    {
        var computer = new Computer();
        
        var cpu = new CPU("Intel i7", 30000, "Intel", 8, 4.7);
        var ram = new RAM("DDR4", 8000, "Kingston", 16, "DDR4");
        var gpu = new GPU("RTX 3080", 60000, "NVIDIA", 10, "PCI-E");
        var hdd = new HDD("SSD", 5000, "Samsung", 1000, "2.5");
        
        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        computer.AddComponent(gpu);
        computer.AddComponent(hdd);
        
        return computer;
    }

}


// ВЫВОД КОДА

// создание компьютеров через фабрику
Console.WriteLine("-==ИГРОВОЙ КОМПЬЮТЕР==-:");
var gamingPC = ComputerFactory.PC();
gamingPC.ShowComponents();


// демонстрация работы с интерфейсом
Console.WriteLine("\n== УСТАНОВКА КОМПОНЕНТОВ: ==");
List<IInstallable> installableComponents = new List<IInstallable>
{
    new CPU("Intel i5", 15000, "Intel", 6, 4.0),
    new RAM("DDR4", 6000, "HyperX", 16, "DDR4")
};

foreach (var component in installableComponents)
{
    component.Install();
}

// Демонстрация Generic методов
Console.WriteLine("\n==  GENERIC МЕТОДЫ: ==");
var testComputer = new Computer();
var testCPU = new CPU("Test CPU", 5000, "Test", 2, 2.0);
var testRAM = new RAM("Test RAM", 2000, "Test", 4, "DDR3");
var testHDD = new HDD("NVMe", 7000, "WD", 2000, "M.2");
var testGPU = new GPU("RX 6700", 35000, "AMD", 12, "PCI-E");


testComputer.AddComponent(testCPU);
testComputer.AddComponent(testRAM);
testComputer.AddComponent(testGPU);
testComputer.AddComponent(testHDD);
testComputer.ShowComponents();

testComputer.RemoveComponent(testRAM);
testComputer.ShowComponents();

```
